###  3. Perform feature engineering 

```python
# Create new features for frequency.
df['Frequency'] = df.groupby('Customer ID')['Order ID'].transform('count')
```

```python
# create a new feature for  recency (time since last order)
df['Recency']= (pd.to_datetime('today') - pd.to_datetime(df['Delivery_Date'])).dt.days
```

```python
# Creatge a new feature CLV
df['CLV'] = df['Total Revenue'].apply(lambda x: float(x.replace('$', '').replace(',', '')))
```

```python
# Create  average cost unit
df['Avg_Unit_Cost'] = df['Unit Cost'].apply(lambda x: float(x.replace('$', '').replace(',', '')))
```

```python
# Create a customer age feature
# create the DOB to datetime
df['DOB'] = pd.to_datetime(df['Customer_BirthDate'], format='%d%b%Y')

# Calculate age
today = pd.Timestamp('now')
df['Age'] = (today - df['DOB']).dt.days // 360
```

###  Column Transformer and Pipeline

```python
# Define the numerical features
numerical_columns = df_aggr_no_id.select_dtypes(include=['float64', 'int64']).columns
```

```python
# Pipeline for preprocessing 
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_columns)])
```

```python
# Fit and transform the data
df_aggr_preprocessed  = preprocessor.fit_transform(df_aggr_no_id)
```

```python
# Visualize the scaled data
df_aggr_preprocessed
```

```python
# Visualise scale data in a dataframe
columns = ['Frequency', 'Recency', 'CLV', 'Avg_Unit_Cost', 'Age']
scaled_df = pd.DataFrame(df_aggr_preprocessed, columns=columns)
scaled_df.head()
```

### Elbow Method Implementation 

```python
# Elbow method with variation of different K
WCSS = []           # variable to store K

for i in range(2, 20):
    kmeans= KMeans(n_clusters=i, init='k-means++', random_state=0, n_init=10 )
    kmeans.fit(df_aggr_preprocessed)
    WCSS.append(kmeans.inertia_)

plt.plot(range(2, 20), WCSS)
plt.title('The Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.show()
```

### Observation 

The Elbow Method was applied, which plots the WCSS against a range of potential k values. The WCSS decreases rapidly from k=1 to approximately k=4, and reveals a clear bend at around k=5 to k=6. 

### Silhouette method implementation 

```python
for n_clusters in range(2, 20):
    clusterer = KMeans(n_clusters = n_clusters, random_state=10, n_init = 10)
    cluster_labels = clusterer.fit_predict(df_aggr_preprocessed)

    silhouette_avg = silhouette_score(df_aggr_preprocessed, cluster_labels)

    print("For n_clusters =", n_clusters,
          "The average silhouette_score is :", silhouette_avg)
```

### Observation
The Silhouette score peaks at k=2 but shows a peak around k=4 to k=7. Beyond k=7, the score declines

### Perform hierarchical clustering and create a dendogram

```python
Z = linkage(df_aggr_preprocessed, method='average')
# Plot dendrogram
plt.figure(figsize=(10, 6))
dendrogram(
    Z,
    truncate_mode='lastp',  # show only the last p clusters
    p=10,                   # number of clusters to show
    leaf_rotation=90,       
    leaf_font_size=12      
    #show_contracted=True    
)
plt.title("Hierarchical Clustering Dendrogram (last 10 clusters)")
plt.xlabel("Cluster Index")
plt.ylabel("Distance")
plt.show()
```

## observation 
1.	Scenario A: Clear split into 2 groups, long vertical line at the top, connecting two large branches, k=2.
2.	Scenario B: Clear split into 3 groups, the line connects clusters within one and two main branches, k=3.
3.	Scenario C: Clear split into 5 Groups. The vertical line is mixed with shorter and longer, k=5.


### Conclusion for k selection 

All three methods converge unequivocally on the same result. The evidence from computational efficiency (Elbow), cluster quality (Silhouette), and natural data structure (Dendrogram) is consistent; thus, the optimal number of clusters that best represents the data is k = 5.

###  Performing k-means clustering based on the optimum number of k,             

```python
# Initialize the K-Means algorithm
kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)
```

```python
 # Create a pipeline for the data processing
pipeline = Pipeline(steps=[('preprocessor', preprocessor),
                           ('clusterer', kmeans)])
```

```python
# Get the cluster assignments 
cluster_labels = pipeline.predict(df_aggr_no_id)
```

### Dimension reduction with PCA 

```python
# visualize the data 
df_aggr_no_id
```

```python
# Write PCA ppipeline
pca = PCA(n_components=2)
```

```python
# Create a pipeline for the data processing
pca_pipeline = Pipeline(steps=[('preprocessor', preprocessor),
                           ('reduce_dim', pca)])
```

```python
# Fit the pipeline to the data
pca_pipeline.fit(df_aggr_no_id)
```

```python
# Get the cluster assignments 
X_pca = pca_pipeline.fit_transform(df_aggr_no_id)
```

```python
print('PCA data for visualisation', X_pca.shape)
print("Scale data shape", df_aggr_no_id.shape)
```

```python
sub_cluster = df_aggr['cluster_label']
sub_cluster
```

```python
# Visualise the reduced data in 2D.

def plot_embedding(df_aggr_no_id, title, sub_cluster):
    plt.figure(figsize=(10, 8))
    for i in range(len(np.unique(sub_cluster))):
        plt.scatter(df_aggr_no_id[sub_cluster == i, 0],
                    df_aggr_no_id[sub_cluster == i, 1],
                    label=str(i))
    plt.xlabel('Component 1')
    plt.ylabel('Component 2')
    plt.legend()
    plt.title(title)
    plt.grid(True)
    plt.show()

plot_embedding(X_pca, "PCA of customer segmentation with clustering", sub_cluster)
```

### Observation 

The PCA successfully reduces dimensionality while preserving cluster structure. the plot demonstrates effective customer segmentation with clearly defined groups that represent distinct behavioral patterns.
